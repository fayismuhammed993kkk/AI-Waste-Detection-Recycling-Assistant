```python
label_folders = [
    "train/labels",
    "valid/labels",
    "test/labels"
]

counter = Counter()

for folder in label_folders:
    folder_path = os.path.join(DATASET_PATH, folder)

    for file in os.listdir(folder_path):

        if file.endswith(".txt"):

            with open(os.path.join(folder_path, file)) as f:

                for line in f:

                    class_id = int(line.split()[0])

                    counter[class_id] += 1
```

```python
df = pd.DataFrame({
    "Class ID": list(counter.keys()),
    "Class Name": [names[i] for i in counter.keys()],
    "Objects": list(counter.values())
})

df = df.sort_values("Objects", ascending=False)

df
```

```python
import shutil

SOURCE = r"C:\Users\fayis\Data Science\Techolas\DLP\Ai_waste_seggregation\Dataset"
DEST = r"C:\Users\fayis\Data Science\Techolas\DLP\Ai_waste_seggregation\Dataset_Optimized"

splits = ["train", "valid", "test"]

for split in splits:

    # Create folders
    (DEST / split / "images").mkdir(parents=True, exist_ok=True)
    (DEST / split / "labels").mkdir(parents=True, exist_ok=True)

    src_img = SOURCE / split / "images"
    dst_img = DEST / split / "images"

    count = 0

    for img in src_img.iterdir():

        if img.is_file():
            shutil.copy2(img, dst_img / img.name)
            count += 1

    print(f"{split}: {count} images copied")

print("\n Image copy completed!")
```

```python
splits = ["train", "valid", "test"]
total_files = 0
total_objects = 0

for split in splits:

    src_labels = SOURCE / split / "labels"
    dst_labels = DEST / split / "labels"

    converted = 0

    for txt_file in src_labels.glob("*.txt"):

        new_lines = []

        with open(txt_file, "r") as f:

            for line in f:

                parts = line.strip().split()

                if len(parts) == 0:
                    continue

                old_class = int(parts[0])

                new_class = class_mapping[old_class]

                parts[0] = str(new_class)

                new_lines.append(" ".join(parts))

                total_objects += 1

        with open(dst_labels / txt_file.name, "w") as f:
            f.write("\n".join(new_lines))

        converted += 1

    total_files += converted

    print(f"{split}: {converted} label files converted")

print("\n===================================")
print(f"Total Label Files : {total_files}")
print(f"Total Objects     : {total_objects}")
print("✅ Label conversion completed!")
```

```python
#VERIFY DATASET
from PIL import Image

splits = ["train", "valid", "test"]

total_images = 0
total_labels = 0
missing_labels = 0
missing_images = 0
invalid_labels = 0
corrupt_images = 0

for split in splits:

    img_dir = DEST / split / "images"
    lbl_dir = DEST / split / "labels"

    images = {f.stem: f for f in img_dir.iterdir() if f.is_file()}
    labels = {f.stem: f for f in lbl_dir.glob("*.txt")}

    total_images += len(images)
    total_labels += len(labels)

    # Missing labels
    for stem in images:
        if stem not in labels:
            missing_labels += 1

    # Missing images
    for stem in labels:
        if stem not in images:
            missing_images += 1

    # Check labels
    for label_file in labels.values():

        with open(label_file) as f:

            for line in f:

                if line.strip():

                    cls = int(line.split()[0])

                    if cls < 0 or cls >= len(new_classes):
                        invalid_labels += 1

    # Check image corruption
    for img in images.values():

        try:
            Image.open(img).verify()
        except:
            corrupt_images += 1

print("="*40)
print("DATASET VERIFICATION")
print("="*40)

print(f"Images            : {total_images}")
print(f"Labels            : {total_labels}")
print(f"Missing Labels    : {missing_labels}")
print(f"Missing Images    : {missing_images}")
print(f"Invalid Labels    : {invalid_labels}")
print(f"Corrupt Images    : {corrupt_images}")

if (
    missing_labels==0 and
    missing_images==0 and
    invalid_labels==0 and
    corrupt_images==0
):
    print("\n🎉 DATASET VERIFIED SUCCESSFULLY!")
else:
    print("\n⚠️ Dataset has issues.")
```
